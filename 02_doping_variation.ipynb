# Effect of Doping Concentration on PN Junction I-V Behaviorm
Extends the Shockley equation model to compare three doping levels by varying I_s.

Higher doping concentrations → lower I_s → diode requires more forward voltage to turn on.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os
os.makedirs("figures", exist_ok = True)
```

```python
V_T = 0.02585 # thermal voltage at room temp (300K)
V = np.linspace(-1, 0.8, 1000) # 1000 even spaces between -1V and 0.8V
```

```python
doping_levels = [
    (1e-9, "Light doping (I_s = 1e-9)"),
    (1e-12, "Moderate doping (I_s = 1e-12)"),
    (1e-15, "Heavy doping (I_s = 1e-15)"),]
```

```python
plt.figure(figsize = (8,5))

# loop through each of the doping levels and plot them on the graph
for I_s, label in doping_levels:
    I = I_s * (np.exp(V/V_T) - 1)
    plt.plot(V, I * 1000, label = label) # plot and label the line with the respective doping level

plt.xlabel("Voltage (V)")
plt.ylabel("Current (mA)")
plt.title("Effect of Doping Concentration on PN Junction I-V Curve")
plt.axhline(0, color = 'black', linewidth = 0.8)
plt.axvline(0, color = 'black', linewidth = 0.8)
plt.ylim(-0.1, 5) # to show all three curves and compare them
plt.legend() # display the label for the respective doping level
plt.grid(True)
plt.tight_layout()
plt.savefig("figures/doping_variation.png", dpi = 150)
plt.show()
```
