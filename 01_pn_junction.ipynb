# PN Junction I-V Characteristics
Simulates the current-voltage (I-V) behavior on a silicon PN Junction using the Shockley diode equation: 
**I = I_s × (e^(V/V_T) - 1)**

The exponential relationship captures the asymmetric behavior of a diode - conducting heavily in forward bias and blocking current in reverse bias.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os
os.makedirs("figures", exist_ok = True)
```

```python
I_s = 1e-12 # reverse saturation current typical for silicon (1 picoamp)
V_T = 0.02585 # thermal voltage at room temp (300K)
```

```python
V = np.linspace(-1, 0.8, 1000) # 1000 even spaces between -1V and 0.8V
```

```python
I = I_s * (np.exp(V/V_T) - 1) # The Shockely equation
```

```python
plt.figure(figsize = (8,5)) 
plt.plot(V, I * 1000) # multiply by 1000 to convert A to mA
plt.xlabel("Voltage (V)")
plt.ylabel("Current (mA)")
plt.title("PN Junction I-V Curve")
plt.axhline(0, color = 'black', linewidth = 0.8) # horizontal line at 0
plt.axvline(0, color = 'black', linewidth = 0.8) # vertical line at 0
plt.ylim(-0.1, 5) # zoom where reverse bias region is visible
plt.grid(True)
plt.tight_layout()
plt.savefig("figures/iv_curve.png", dpi = 150)
plt.show()
```
